```python
# Import pandas
import pandas as pd
```

```python
# Create a Series of three different colours
colours = pd.Series(['Red', 'Blue', 'Yellow'])
```

```python
# View the Series of different colours
colours
```

```python
# Create a Series of three different car types and view it
car_types = pd.Series(["Toyota", "BMW", "Nissan"])
car_types
```

```python
# Combine the Series of cars and colours into a DataFrame
df = pd.DataFrame(
  {
    'Car Type': car_types,
    'Colour': colours
	}
)
df
```

```python
# Select the "Odometer (KM)" column from the car sales DataFrame
car_sales['Odometer (KM)']
```

```python
# Find the mean of the "Odometer (KM)" column in the car sales DataFrame
mean_odometer = car_sales['Odometer (KM)'].mean()
print(f"'Odometer (KM)' Column Mean: {mean_odometer}")
```

```python
# Select the rows with over 100,000 kilometres on the odometer
car_sales[car_sales['Odometer (KM)'] > 100000]
```

```python
# Group columns of the car sales DataFrame by the Make column and find the average
car_sales.groupby(['Make']).mean(numeric_only=True)
```

```python
# Import Matplotlib and create a plot of the Odometer column
# Don't forget to use %matplotlib inline
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(layout='constrained')
ax.plot(car_sales['Odometer (KM)'])
ax.set_title('Plot of Odometer column:')

plt.show()
```

```python
# Create a histogram of the Odometer column using hist()
fig, ax = plt.subplots(layout='constrained')
ax.hist(car_sales['Odometer (KM)'])
ax.set_title('Histogram of Odometer column:')

plt.show()
```
